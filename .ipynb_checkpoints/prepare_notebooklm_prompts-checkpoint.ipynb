```python
import argparse
from pathlib import Path
from datetime import datetime
import csv
import openpyxl
```

```python
DEFAULT_NOTEBOOK_NAME = "FuturesSummit_2025_Qual_Analysis_AllSessions"

PASS_BUCKETS = {
    "PASS_1": ["Key_Learnings", "Cross_Cutting_Themes", "Recurring_Audience_Questions"],
    "PASS_2": ["Missing_Elements", "Unmet_Needs"],
    "PASS_3": ["Use_Cases_Exemplars"],
}
```

```python
def make_run_id(now=None):
    now = now or datetime.now()
    return f"RUN_{now:%Y%m%d_%H%M}"

def extract_prompt_from_sheet(ws):
    # Find row containing "NotebookLM Prompt"
    prompt_label_row = None
    for r in range(1, 60):
        v = ws.cell(r, 1).value
        if isinstance(v, str) and v.strip().lower() == "notebooklm prompt":
            prompt_label_row = r
            break
    if not prompt_label_row:
        raise ValueError(f'Cannot find "NotebookLM Prompt" in sheet {ws.title}')

    # Prompt starts on next row, col A, continues until a blank line or the header row (~row 11)
    prompt_lines = []
    for r in range(prompt_label_row + 1, prompt_label_row + 25):
        v = ws.cell(r, 1).value
        if v is None:
            if prompt_lines:
                break
            else:
                continue
        txt = str(v).strip()
        if not txt:
            if prompt_lines:
                break
            continue
        # Stop if we hit the table header row (contains *_ID in first cell)
        if txt.endswith("_ID") or txt.endswith("ID"):
            break
        prompt_lines.append(txt)

    if not prompt_lines:
        raise ValueError(f"No prompt content found under 'NotebookLM Prompt' in {ws.title}")

    return "\n\n".join(prompt_lines).strip()
```

```python
def write_prompt_packet(out_path: Path, run_id: str, pass_id: str, bucket: str,
                        notebook_name: str, analyst: str, prompt_text: str):
    header = f"""RUN_ID: {run_id}
PASS_ID: {pass_id}
BUCKET: {bucket}
NOTEBOOK: {notebook_name}
DATE: {datetime.now():%Y-%m-%d}
ANALYST: {analyst}
SOURCE_SET: All transcripts in 02_Transcripts_Clean
---BEGIN_PROMPT_TO_PASTE_IN_NOTEBOOKLM---
"""
    footer = """
---END_PROMPT_TO_PASTE_IN_NOTEBOOKLM---
"""
    # Force numbered output for deterministic parsing later
    format_rule = """
Response format requirement:
Return your output as a numbered list (1., 2., 3., ...), one finding per number.
For each numbered item, include session IDs in parentheses when possible (e.g., FS01, FS07).
""".strip()

    content = header + prompt_text.strip() + "\n\n" + format_rule + "\n" + footer
    out_path.write_text(content, encoding="utf-8")
```

```python
def update_session_index(session_index_path: Path, run_id: str, pass_id: str, status: str):
    rows = []
    with session_index_path.open("r", encoding="utf-8", newline="") as f:
        reader = csv.DictReader(f)
        fieldnames = list(reader.fieldnames or [])
        for row in reader:
            rows.append(row)

    # Add audit fields if missing
    for new_col in ["Last_Run_ID", "Last_Run_Pass"]:
        if new_col not in fieldnames:
            fieldnames.append(new_col)

    # Update all rows (since NotebookLM notebook is cross-session)
    for row in rows:
        row["NotebookLM_Status"] = status
        row["Last_Run_ID"] = run_id
        row["Last_Run_Pass"] = pass_id

    with session_index_path.open("w", encoding="utf-8", newline="") as f:
        writer = csv.DictWriter(f, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(rows)
```

```python
def main():
    ap = argparse.ArgumentParser()
    ap.add_argument("--framework", required=True, help="Path to Qual_Analysis_Framework.xlsx")
    ap.add_argument("--project_root", required=True, help="Project root folder")
    ap.add_argument("--session_index", required=True, help="Path to Session_Index.csv")
    ap.add_argument("--pass_id", default="PASS_1", choices=PASS_BUCKETS.keys())
    ap.add_argument("--notebook_name", default=DEFAULT_NOTEBOOK_NAME)
    ap.add_argument("--analyst", default="Hongling")
    args = ap.parse_args()

    framework_path = Path(args.framework)
    project_root = Path(args.project_root)
    session_index_path = Path(args.session_index)

    run_id = make_run_id()
    pass_id = args.pass_id

    inbox_dir = project_root / "03_NotebookLM_Exports" / "INBOX"
    inbox_dir.mkdir(parents=True, exist_ok=True)

    wb = openpyxl.load_workbook(framework_path)
    buckets = PASS_BUCKETS[pass_id]

    written = []
    for bucket in buckets:
        if bucket not in wb.sheetnames:
            raise ValueError(f"Sheet '{bucket}' not found in {framework_path.name}")
        ws = wb[bucket]
        prompt_text = extract_prompt_from_sheet(ws)

        out_file = inbox_dir / f"{run_id}__{bucket}__PromptPacket.md"
        write_prompt_packet(out_file, run_id, pass_id, bucket, args.notebook_name, args.analyst, prompt_text)
        written.append(out_file.name)

    # Log status as "prepared" (we haven’t run prompts yet)
    update_session_index(session_index_path, run_id, pass_id, status=f"{pass_id} prepared")

    print(f"RUN_ID: {run_id}")
    print(f"PASS_ID: {pass_id}")
    print("WROTE:")
    for w in written:
        print(f" - {w}")

if __name__ == "__main__":
    main()
```
