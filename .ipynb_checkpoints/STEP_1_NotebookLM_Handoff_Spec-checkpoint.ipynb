# Step 1 — NotebookLM → Excel Handoff Specification (SOP)

## Purpose
Establish a consistent, reproducible handoff format so NotebookLM outputs can be parsed by Python and registered into `Qual_Analysis_Framework.xlsx`, with status logged in `Session_Index.csv`.

## Artifacts
* **Qual_Analysis_Framework.xlsx**: prompts + authoritative outputs
* **Session_Index.csv**: status tracker
* **03_NotebookLM_Exports/INBOX/**: staging area for NotebookLM responses

## Run IDs
* **Format**: `RUN_YYYYMMDD_HHMM`
* **Example**: `RUN_20260116_1430`

## INBOX Response Files
* **Location**: `03_NotebookLM_Exports/INBOX/`
* **Naming**: `<RunID>__<BucketName>__NotebookLM_Response.md`

## Response Header Block (Required)
Paste this at the top of every INBOX response file:

```text
RUN_ID: <RUN_ID>
PASS_ID: <PASS_ID>
BUCKET: <BucketName>
NOTEBOOK: FuturesSummit_2025_Qual_Analysis_AllSessions
DATE: <YYYY-MM-DD>
ANALYST: <Name>
SOURCE_SET: All transcripts in 02_Transcripts_Clean
---BEGIN_NOTEBOOKLM_RESPONSE---
<paste NotebookLM response here>
---END_NOTEBOOKLM_RESPONSE---
